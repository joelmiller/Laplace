# Review of pre-requisite skills

If you've ever played a sport, you probably know that sometimes you can get away with not mastering some skill at one level, but it is essential to succeed against higher competition.  You also know that to master a skill, you need repetition.  It's not enough to watch the coach demonstrate a technique and then move on.  It's definitely not enough to play a video game where your digital avatar can do it automatically.

You have to do it yourself until you master it.  Then you have to use the technique in situations other than the controlled cases that you've practiced.

Sometimes an athlete moves up to a higher level and discovers that they are missing skills they were expected to already learn.  To succeed, they need to do extra work to master the missing skills while simultaneously working on the new things their coach is teaching.  This is very difficult.

Many students who arrive in Vector Calculus but are missing important Algebra or single variable Calculus skills.  They need to master these while learning the new Vector Calculus material.  Having their calculator master it is not enough.  This is not easy.  If you are in this situation, you should allocate extra time beyond the standard amount which is expected of students who do not need to review the prerequisites.
