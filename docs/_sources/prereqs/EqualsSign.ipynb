# The use of $=$ and $\Rightarrow$

Mathematical writing has grammar rules, though they are not frequently explicitly taught.  Some of the most common errors occur when people misuse either $=$ or $\Rightarrow$.

In mathematical writing, the equals sign takes a specific meaning.  Unfortunately, it is different from what the equals sign means in primary school and what it means with a calculator.  This leads to confusion.

In primary school, you would see something like $3+4=~~~$ and you would be expected to fill in $7$.  For most calculators, the equals sign plays a similar role. You type the keys in order: $3$, $+$, $4$, $=$, and then the calculator displays $7$.  So many students are used to the idea that the equals sign is the instruction "perform the calculation".  This is not what it means.

In mathematical writing, the equals sign sits in the middle of an equation.  An equation such as $A=B$ is an assertion or claim that the two things on either side of the equals sign are in fact equal to one another.  

Consider now $\Rightarrow$.  This is used to denote that if one statement is True, then the next statement must also be True.  So $A=B \Rightarrow 2A=2B$.  If the first equation is True, then the second equation is also True.  Most mathematical writing of the variety that shows up in MAT2VCA should be presented as a sequence of equations, each of which has $=$ in the middle and between the equations we use $\Rightarrow$.

Let us consider an illustrative example showing two correct uses of $=$ and $\Rightarrow$ and then an incorrect use.  They all demonstrate correct reasoning and get to the same answer.  However, the first two use mathematical grammar correctly.  The third is more like a string of incomplete sentences.

Here is a sample problem: *Consider $f(x) = x^2 -6x + 9$.  Find $f(3)$.*

- Here is the first way:
  \begin{align*}
  f(x) &= x^2 - 6x + 9\\
  \Rightarrow f(3) &= 3^2 - 6\times 3 + 9\\
  \Rightarrow f(3) &= 9 - 18 + 9\\
  \Rightarrow f(3) &= 0
  \end{align*}

- A shortcut we use in mathematics when the the left hand side of a sequence of equations remains the same is to write a sequence of equals signs, without including the left hand side:
  \begin{align*}
  f(x) &= x^2-6x+9\\
  \Rightarrow f(3) &= 3^2-6\times 3 + 9\\
  &= 9 - 18 + 9\\
  &= 0
  \end{align*}


- Here is an incorrect way of writing the solution.  It matches exactly the sequence of steps you would use if using a (standard) calculator.  The thought process underlying these steps is correct.  However, the presentation is incorrect:

  $f(3)=3^2 = 9 - 6\times 3 = 9-18=-9+9=0$


Notice that in the first two examples, the mathematical calculation happens **between** each line.  There is a specific calculation that is done from one equation to the next.  If that calculation is done correctly and one equation is True, then the equation on the next line is also True.

In the final example, you can probably determine the steps that are being taken.  But as written, this is the equivalent of writing an essay that is full of grammatically incorrect sentences and lacking periods.  For example, we have the second equals sign which gives us the equation:

$$
3^2 = 9 - 6\times 3
$$
but the writer means "$3^2$ equals $9$, and then the next thing we will do is subtract $6\times 3$ from this."  There are two statements mixed together with this one equals sign.  

This is a problem because it's hard to know what each equals sign really means.  In the "correct" versions higher up, I can ask you to verify the step between the second and third line and you can check exactly what calculation was done and whether it was done correctly.  You need to make sure sure that if the first equation is true, then the next equation is also true.

However, as this is written, if I ask you to check the fourth equals sign ($9-18=-9+9$), you can't do that in isolation.  You have to know that the writer means "in this step I find $9-18 =-9$, and now I'm starting the next step which is adding $9$."  We can figure that out, but what if the writer makes a mistake and has $9-18=9+9$?  Now you have to figure out what was going on in the writer's head - did the writer make a mistake?  Was the writer doing a different step of the calculation without being clear?


## Summary

The basic structure of any mathematical writing is that we start from a set of claims that we assume to be true, and then we follow a series of logical steps which lead to a conclusion.  At each step in our work, we make a new claim.  In mathematical writing we must make sure that our claim is clearly stated and that it is also clear how one claim follows from the previous claim.

For the sort of problems we work on in this subject, our claims are typically expressed through equations.  We claim that the expression on the left hand side of $=$ is exactly the same as on the right hand side.  The $\Rightarrow$ means that if the previous equation is True, then so is the next one.  We use the word "equals" when we read $=$ and "implies" when we read $\Rightarrow$.

One of the skills we focus on in this subject is ensuring that students can clearly express a mathematical explanation.  If you write down an equals sign and the two things on either side of it are not equal, you will lose points, even if the underlying calculation that you are doing is correct.
